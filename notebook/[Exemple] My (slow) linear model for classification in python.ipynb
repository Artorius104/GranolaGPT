```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
points = np.random.random((100, 2)) * 100.0
```

```python
plt.scatter(points[:, 0], points[:, 1])
plt.show()
```

```python
colors = []
y_targets = []
for p in points:
  if p[0] + 3.0 * p[1] - 78.0 * 3 >= 0:
    colors.append('blue')
    y_targets.append(1.0)
  else:
    colors.append('red')
    y_targets.append(-1.0)
```

```python
plt.scatter(points[:, 0], points[:, 1], c=colors)
plt.show()
```

```python
W = np.random.random(3) * 2.0 - 1.0
```

```python
W
```

```python
background_points = []
background_colors = []
for x1 in range(0,100):
  for x2 in range(0, 100):
    weighted_sum = W[0] * 1.0 + W[1] * x1 + W[2] * x2
    background_points.append([x1, x2])
    if weighted_sum >= 0:
      background_colors.append('lightblue')
    else:
      background_colors.append('pink')
background_points = np.array(background_points)
```

```python
plt.scatter(background_points[:, 0], background_points[:, 1], c=background_colors)
plt.scatter(points[:, 0], points[:, 1], c=colors)
plt.show()
```

```python
for it in range(1000000):
  k = np.random.randint(0, len(points))
  yk = y_targets[k]
  Xk = np.array([1.0, *points[k]])
  weighted_sum = (W * Xk).sum()
  gXk = 1.0 if weighted_sum >= 0 else -1.0
  W += 0.001 * (yk - gXk) * Xk
```

```python
background_points = []
background_colors = []
for x1 in range(0,100):
  for x2 in range(0, 100):
    weighted_sum = W[0] * 1.0 + W[1] * x1 + W[2] * x2
    background_points.append([x1, x2])
    if weighted_sum >= 0:
      background_colors.append('lightblue')
    else:
      background_colors.append('pink')
background_points = np.array(background_points)
```

```python
plt.scatter(background_points[:, 0], background_points[:, 1], c=background_colors)
plt.scatter(points[:, 0], points[:, 1], c=colors)
plt.show()
```
