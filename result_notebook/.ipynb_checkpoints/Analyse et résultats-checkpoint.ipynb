# MLP : Résultats et observations sur le dataset

```python
import os
import sys
import io
import ctypes
import json
from ctypes import *
import numpy as np
import numpy.ctypeslib
from PIL import Image

from utils import plot_accuracies, plot_losses, load_json
from dataset import load_image, load_dataset, shuffle_dataset, z_normalize, display_images

from lib_loader import load_my_lib

seed = 42
```

## Dataset

```python
# Charger le dataset
image_size = 48  # Taille des images
dataset_path = '../dataset_upgrade'
X_train, y_train, X_test, y_test = load_dataset(dataset_path, image_size)
X_train, y_train = shuffle_dataset(X_train, y_train)
X_test, y_test = shuffle_dataset(X_test, y_test)

print(f"X_train shape: {X_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_test shape: {y_test.shape}")

display_images(X_train, 10)
```

## Chargement de la librairie

```python
path = "../my_lib/target/release/deps/libmy_lib.so"

lib = load_my_lib(path)
```

### Preprocessing des données

```python
# Normaliser les données
X_train_normalized, X_test_normalized = z_normalize(X_train, X_test)

# Aplatir les images pour MLP
X_train_flat = X_train_normalized.reshape(X_train_normalized.shape[0], -1)
X_test_flat = X_train_normalized.reshape(X_train_normalized.shape[0], -1)

input_dim = X_train_flat.shape[1]
output_dim = y_train.shape[1]
```

## Pipeline d'entraînement

### Modèle

```python
layer_sizes = np.array([input_dim, 16, output_dim], dtype=np.int32)
num_layers = len(layer_sizes)

# Créer le modèle MLP
mlp = lib.create_MyMLP(layer_sizes, num_layers, seed)

# Afficher la summary
lib.summary_MyMLP(mlp)
```

### Entraînement

```python
# Paramètres d'entraînement
alpha = 0.0001
epochs = 50
is_classification = True

# Entraîner le modèle
lib.train_MyMLP(
    mlp,
    X_train_flat,
    y_train,
    X_test_flat,
    y_test,
    X_train_flat.shape[0],
    X_test_flat.shape[0],
    X_train_flat.shape[1],
    y_train.shape[1],
    alpha,
    epochs,
    is_classification
)
```

### Evaluation

```python
# Tester le modèle
predictions = []
for sample in X_test_flat:
    sample = np.array(sample, dtype=np.float64)
    output = lib.predict_MyMLP(mlp, sample, 1, sample.size, is_classification)
    prediction = [output[i] for i in range(y_train.shape[1])]
    predictions.append(prediction)

predictions = np.array(predictions)

# Convertir les prédictions en classes
predicted_classes = np.argmax(predictions, axis=1)
true_classes = np.argmax(y_test, axis=1)

# Calculer l'exactitude
accuracy = np.mean(predicted_classes == true_classes)
print(f'Accuracy: {accuracy * 100:.2f}%')

# Afficher quelques résultats
for i in range(10):
    print(f"True: {true_classes[i]}, Predicted: {predicted_classes[i]}")
```

### Libération de la mémoire

```python
# Libérer la mémoire
lib.destroy_MyMLP(mlp)
```

## Structure simple

Commençons très simplement en ajoutant une seule couche cachée afin de se faire une première idée de l'impact d'une couche.

### Focus sur le learning rate

```python
path = '../PA_models/MLP_models/lr_00001_6912_8_3/metrics.json'

history = load_json(path)

train_losses = history["train_losses"]
test_losses = history["test_losses"]

train_accuracies = history["train_accuracies"]
test_accuracies = history["test_accuracies"]

plot_losses(train_losses, test_losses)
plot_accuracies(train_accuracies, test_accuracies)
```

```python
path = '../PA_models/MLP_models/lr_0001_6912_8_3/metrics.json'

history = load_json(path)

train_losses = history["train_losses"]
test_losses = history["test_losses"]

train_accuracies = history["train_accuracies"]
test_accuracies = history["test_accuracies"]

plot_losses(train_losses, test_losses)
plot_accuracies(train_accuracies, test_accuracies)
```

Les courbes de perte montrent que le modèle s'ajuste bien aux données, avec des pertes d'entraînement et de test très basses et similaires. Cependant, les courbes de précision révèlent un possible sous-ajustement, avec des précisions relativement faibles (0.45 pour l'entraînement et 0.42 pour le test) et un écart constant entre les deux.

- **Underfitting** Ces résultats semblent nous dire que le modèle ne s'est pas entraîné assez longtemps.
- **Fluctuations :** Nous avons deux hypothèses : soit le taux d'apprentissage est trop élevé, soit il y a du bruit au sein des données. Nous avons constaté que même avec un taux d'apprentissage très faible, les fluctuations persistent, sans toutefois altérer la forme générale de la courbe. Ainsi, nous penchons plutôt pour la deuxième hypothèse, étant donné que nous traitons des images.

```python
path = '../PA_models/MLP_models/lr_001_6912_8_3/metrics.json'

history = load_json(path)

train_losses = history["train_losses"]
test_losses = history["test_losses"]

train_accuracies = history["train_accuracies"]
test_accuracies = history["test_accuracies"]

plot_losses(train_losses, test_losses)
plot_accuracies(train_accuracies, test_accuracies)
```

#### Ajoutons des neurones

Depuis les observations des courbes précédentes, nous pouvons déjà deviner que ajouter des neurones à la couche cachée rique d'accentuer l'effet d'overfitting.

```python
path = '../PA_models/MLP_models/lr_001_6912_32_3/metrics.json'

history = load_json(path)

train_losses = history["train_losses"]
test_losses = history["test_losses"]

train_accuracies = history["train_accuracies"]
test_accuracies = history["test_accuracies"]

plot_losses(train_losses, test_losses)
plot_accuracies(train_accuracies, test_accuracies)
```

- **Perte d'entraînement :** Diminue régulièrement et atteint un niveau bas.
- **Perte de test :** Diminue initialement, puis se stabilise à un niveau plus élevé.
- **Accuracy d'entraînement :** Augmente et atteint environ 70%.
- **Accuracy de test :** Augmente puis se stabilise autour de 55%.

#### Déductions :
- **Overfitting :** L'écart entre les courbes d'entraînement et de test indique un sur-ajustement du modèle.
